# **Operators**

```python
# Arithmetic Operators ( + , - , * , / , % , ** , // )
print("Addition", "7 + 3 = ", 7 + 3)
print("Subtraction", "7 - 3 = ", 7 - 3)
print("Multiplication", "7 * 3 = ", 7 * 3)
print("division", "7 / 3 = ", 7 / 3)
print("Modulus", "7 % 3 = ", 7 % 3)
print("Exponentiation", "7 ** 3 = ", 7 ** 3)
print("Floor division", "7 // 3 = ", 7 // 3)
```

```python
# Comparison (Relational) Operators ( == , != , > , < , <= , >= )
print("Equal", "7 == 3 : ", 7 == 3)
print("Not Equal", "7 != 3 : ", 7 != 3)
print("Greater", "7 > 3 : ", 7 > 3)
print("Less", "7 < 3 : ", 7 < 3)
print("Greater or Equal", "7 >= 3 : ", 7 >= 3)
print("Less or Equal", "7 <= 3 : ", 7 <= 3)
```

```python
# Assignment Operators ( = , += , -+ , *= , /= , %= , ** , //= )

# Assigning
v1 = 6
print("v1 = ",v1)

v1+=7
print("v1+=4 = ",v1)

v1-=9
print("v1-=4 = ",v1)

v1*=2
print("v1*=3 = ",v1)

v1/=3
print("v1/=2 = ",v1)

v1%=1
print("v1%=2 = ",v1)

v1**=2
print("v1**=2 = ",v1)

v1//=4
print("v1//=3 = ",v1)
```

```python
# Logical Operators ( not , and , or )
print("================= AND =================")
print("True and True : ", True and True)
print("True and False : ", True and False)
print("False and True : ", False and True)
print("False and False : ", False and False)

print("================= OR =================")
print("True or True : ", True or True)
print("True or False : ", True or False)
print("False or True : ", False or True)
print("False or False : ", False or False)

print("================= NOT =================")
print("not True : ", not True)
print("not False : ", not False)

```

```python


# Bitwise Operators (& , | , ^ , ~ , << , >> )

a = 60            # 60 = 0011 1100
b = 13            # 13 = 0000 1101

c = a & b;        # 12 = 0000 1100
print ("and ", "a & b :  ", c)

c = a | b;        # 61 = 0011 1101
print ("or ", "a | b :  ", c)

c = a ^ b;        # 49 = 0011 0001
print ("xor ", "a ^ b :  ", c)

c = ~b;           # -61 = 1100 0011
print("not ", "~a :  ", c)

c = a << 3;       # 240 = 1111 0000 |||||
print ("shift left ", "a << 2 :  ", c)

c = a >> 3;       # 15 = 0000 1111   ||||| 1000011 1
print ("shift right ","a >> 2 :  ", c)
```

```python
# Create function with 1 argument
def getStudentAge(sName):

  studentsAges = {
   "mahmoud" : 25,
   "ahmad" : 22,
   "khaled" : 28,
   "rula" : 25,
   }

  if sName in studentsAges.keys():
     return studentsAges[sName]

  return "student is not exist"
```

```python
# Create function where arguments with default value
def powNum(num, p=2):
  val= num
  while p !=1:
    val *= num
    p -= 1

  return val
```

```python
print(powNum(4))
```

```python
# Crate function with variable-length argument
def getAverage(*values):
  sum = 0
  count = 0
  for v in values:
    sum+=v
    count+=1

  return sum/count
```

```python
print(getAverage(2,3,4,5,6,7))
```
